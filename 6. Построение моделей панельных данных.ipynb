### Сравнение Random Effects и OLS

В целях выбора между обычной МНК (Pooled model) и моделью со случайными эффектами проводим тест Бреуша-Пагана. Выясним, есть ли случайные эффекты.

H0: нет индивидуальных эффектов (то есть случайные эффекты для всех объектов в выборке одинаковые)

```python
from statsmodels.stats.diagnostic import het_breuschpagan
residuals_pooled_OLS = pooled_res.resids #сохраняем остатки
pooled_OLS_dataset = pd.concat([dfn, residuals_pooled_OLS], axis=1) #добавляем остатки в исходный датасет
breusch_pagan_test_results = het_breuschpagan(pooled_OLS_dataset['residual'], exog)
print('P-value:', breusch_pagan_test_results[3])
```

```python
print(5.502950812785876e-08<0.05)
```

p-value ниже 0,05, значит гипотеза H0 отвергается. Можно сделать вывод о наличии индивидуальных эффектов. Значит нужна модель с эффектами (случайными или фиксированными).

### Выбор между моделью с фиксированными эффектами или моделью со случайными эффектами

Проводим тест Хаусмана.

H0: нет корреляции между индивидуальными эффектами и регрессорами

```python
import numpy.linalg as la
from scipy import stats
def hausman(fe, re):
    b = fe.params
    B = re.params
    v_b = fe.cov
    v_B = re.cov
    df = b[np.abs(b) < 1e8].size
    chi2 = np.dot((b - B).T, la.inv(v_b - v_B).dot(b - B))
    pval = stats.chi2.sf(chi2, df)
    return chi2, df, pval

hausman_results = hausman(fe_res, re_res)
print('p-value: ' + str(hausman_results[2]))
```

```python
print(8.682361239180854e-31<0.05)
```

p-value теста ниже уровня значимости 0.05. Нулевая гипотеза отвергается, значит, есть корреляция между регрессорами и индивидуальными эффектами. Модель со случайными эффектами в таком случае не подходит, поэтому выбираю модель с фиксированными эффектами (Fixed effects model).

###Сравнение моделей и выбор между ними

####Random Effects и OLS

Проведу тест Бреуша-Пагана.

H0: нет индивидуальных эффектов (то есть случайные эффекты для всех объектов в выборке одинаковые)

```python
from statsmodels.stats.diagnostic import het_breuschpagan
residuals_pooled_OLS = pooled_res2.resids #сохраняем остатки
pooled_OLS_dataset = pd.concat([df2, residuals_pooled_OLS], axis=1) #добавляем остатки в исходный датасет
breusch_pagan_test_results = het_breuschpagan(pooled_OLS_dataset['residual'], exog2)
print('P-value:', breusch_pagan_test_results[3])
```

p-value ниже 0,05, значит гипотеза H0 отвергается. Можно сделать вывод о наличии индивидуальных эффектов. Значит нужна модель с эффектами (случайными или фиксированными).

####Fixed effects model или Random effects model

Проведу тест Хаусмана.

H0: нет корреляции между индивидуальными эффектами и регрессорами

```python
import numpy.linalg as la
from scipy import stats
def hausman(fe, re):
    b = fe.params
    B = re.params
    v_b = fe.cov
    v_B = re.cov
    df = b[np.abs(b) < 1e8].size
    chi2 = np.dot((b - B).T, la.inv(v_b - v_B).dot(b - B))
    pval = stats.chi2.sf(chi2, df)
    return chi2, df, pval

hausman_results = hausman(fe_res2, re_res2)
print('p-value: ' + str(hausman_results[2]))
```

p-value теста ниже уровня значимости 0.05. Нулевая гипотеза отвергается, значит, есть корреляция между регрессорами и индивидуальными эффектами. Модель со случайными эффектами в таком случае не подходит, поэтому выбираю модель с фиксированными эффектами (Fixed effects model).
